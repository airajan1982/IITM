## 2. Load a Small LLM and Generate Three Continuations

We will use **DistilGPT-2**, a smaller GPT-2 model suitable for experimentation.

Prompt:

> AI is transforming industries by ...

```python
from transformers import pipeline

# Create a text-generation pipeline
generator = pipeline(
    "text-generation",
    model="distilgpt2"
)

prompt = "AI is transforming industries by"

# Generate three different continuations
outputs = generator(
    prompt,
    max_new_tokens=40,
    num_return_sequences=3,
    do_sample=True,
    temperature=0.8,
    top_p=0.95,
    pad_token_id=generator.tokenizer.eos_token_id
)

for i, output in enumerate(outputs, start=1):
    print(f"Continuation {i}:")
    print(output["generated_text"])
    print("-" * 80)
```

## 3. Tokenization Demo

Sentence:

> LLMs are powerful tools for natural language understanding.

We will tokenize the sentence using the same DistilGPT-2 tokenizer and display:

- Tokens
- Token IDs
- Sequence length

```python
from transformers import AutoTokenizer

sentence = "LLMs are powerful tools for natural language understanding."

# Load the tokenizer associated with DistilGPT-2
tokenizer = AutoTokenizer.from_pretrained("distilgpt2")

# Encode the sentence
encoded = tokenizer(sentence, add_special_tokens=True)

# Get token IDs
token_ids = encoded["input_ids"]

# Convert token IDs back to readable token strings
tokens = tokenizer.convert_ids_to_tokens(token_ids)

print("Sentence:")
print(sentence)

print("\nTokens:")
print(tokens)

print("\nToken IDs:")
print(token_ids)

print("\nSequence length:")
print(len(token_ids))
```

## 4. Display Tokenization as a Table

The next cell makes it easier to understand the relationship between each token and its numerical ID.

```python
# Display token, token ID, and position together

print(f"{'Position':<10}{'Token':<25}{'Token ID':<10}")
print("-" * 45)

for position, (token, token_id) in enumerate(zip(tokens, token_ids)):
    print(f"{position:<10}{repr(token):<25}{token_id:<10}")
```

## 5. Key Concepts

| Concept | Meaning |
|---|---|
| **LLM** | Large Language Model trained to process and generate language |
| **Transformer** | Neural-network architecture behind many modern language models |
| **Tokenizer** | Converts text into tokens that a model can process |
| **Token** | A piece of text such as a word, subword, space-prefixed word, or punctuation |
| **Token ID** | Integer representing a token in the tokenizer vocabulary |
| **Sequence Length** | Number of tokens produced for the input text |
| **Temperature** | Controls randomness/diversity during generation |
| **top_p** | Limits sampling to a probability mass of likely next tokens |
