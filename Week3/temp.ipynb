```python
# In sum we can pass axis=0 to count the number of missing values in each column, or axis=1 to count the number of missing values in each row.
normalized_data.isnull().sum() #isnull().sum() method is used to count the number of missing values in each column of a DataFrame. It returns a Series where the index represents the column names and the values represent the count of missing values in each column.
```
