```python
# importing required libraries
import pandas as pd
import numpy as np
```

```python
# Load sales data from CSV file
path = "D:/Learning/IITM/Week3/sales.csv"
salesData = pd.read_csv(path)
```

```python
# Display the complete sales dataset
salesData
```

```python
# validating null data
salesData.isnull().sum()
```

```python
# Identify duplicate records by order_id
order_id_duplicates = salesData[salesData.duplicated(subset=['order_id'])]
order_id_duplicates
```

```python
# Find orders in which freight cost is high compared to the product price
# A high freight order is one where the shipping cost (freight_value) exceeds the product price
high_freight_orders = salesData[salesData['freight_value'] > salesData['price']]

# Display order_id, product_id, price and freight_value for these orders
high_freight_orders[['order_id', 'product_id', 'price', 'freight_value']]
```

```python
# Find Life time value of each unique customer id. Nagarajan code.
salesData.groupby('customer_id')['payment_value'].sum()
```

```python
# Find Life time value of each unique customer id. AI generated code
# salesData.groupby('customer_id'): Groups the dataset by each unique customer. 
# All rows belonging to the same customer are gathered together
# .['payment_value'].sum(): Selects the payment column and adds up all the purchases made by each individual customer.
# .reset_index(name='lifetime_value'): Converts the grouped data back into a clean, flat DataFrame. It names the newly calculated total column lifetime_value.
# .sort_values(by='lifetime_value', ascending=False): Sorts the final list in descending order, placing your highest-paying customers at the very top.
salesData.groupby('customer_id')['payment_value'].sum().reset_index(name='lifetime_value').sort_values(by='lifetime_value', ascending=False)

```

```python
# How many unique payment types. Nagarajan Code
salesData.groupby('payment_type')['payment_type'].count()
```

```python
# How many order by credit card which > 20$.
salesData[(salesData['payment_type'] == 'credit_card') & (salesData['payment_value'] > 20)].shape[0]
```

```python
# How many orders purchased per each year. AI generated code.
# 1. Convert the column to datetime objects
salesData['order_purchase_timestamp'] = pd.to_datetime(salesData['order_purchase_timestamp'])

# 2. Group by the year and count the unique orders
yearly_orders = salesData.groupby(salesData['order_purchase_timestamp'].dt.year)['order_id'].nunique()

print(yearly_orders)
```
